```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
多平台大盘日报 PDF 生成 + Telegram 发送
"""

import os, sys, glob, requests
import pandas as pd
import numpy as np
from datetime import date
from pathlib import Path
from reportlab.lib import colors
from reportlab.lib.units import mm
from reportlab.platypus import SimpleDocTemplate, Table, TableStyle, Paragraph, Spacer
from reportlab.lib.styles import ParagraphStyle
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.cidfonts import UnicodeCIDFont
from reportlab.pdfbase.ttfonts import TTFont

# ══════════════════════════════════════════════════════
# ★ 多平台配置
# ══════════════════════════════════════════════════════
PLATFORMS = [
    {
        "name":       "oro7x",
        "dept":       "智利",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\oro7x自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8926077105:AAEIGSAowCtBg8TNk6_H-0kez8hqBo8BEx4",
        "chat_id":    "-1003723896850",
        "n_days":     35,
    },
    {
        "name":       "Afun_mx",
        "dept":       "墨西哥",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\Afun_mx自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },
    {
        "name":       "Soluno",
        "dept":       "天龙",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\soluno自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-1003923836069",
        "n_days":     35,
    },
    {
        "name":       "sol777",
        "dept":       "YB",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\sol777自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },
    {
        "name":       "Lucro",
        "dept":       "自营",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\Lucro自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },
    {
        "name":       "3bet",
        "dept":       "A8",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\3bet自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },
    {
        "name":       "Wey7",
        "dept":       "天成",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\wey7自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8992778127:AAHh2aopYmrD9pfnNkOZUg5f5GZKs4If6f4",
        "chat_id":    "-5056696869",
        "n_days":     35,
    },
    {
        "name":       "letjoy",
        "dept":       "",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\letjoy自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },
    {
        "name":       "G777",
        "dept":       "",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\G777自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8866705424:AAGa-0CjpbVJr43ObzlbY32ZLI2ySmeLAL0",
        "chat_id":    "-5281514567",
        "n_days":     35,
    },
    # {
    #     "name":       "Luna777",
    #     "dept":       "",
    #     "channel":    "短信/投放",
    #     "excel_dir":  r"D:\Luna777自动播报",
    #     "excel_glob": "日报-大盘日报_*.xlsx",
    #     "output_dir": r"D:\自动播报报告",
    #     "bot_token":  "8866705424:AAGa-0CjpbVJr43ObzlbY32ZLI2ySmeLAL0",
    #     "chat_id":    "-5281514567",
    #     "n_days":     35,
    # },

    # ── 新增盘：复制下方模板并修改 ──────────────────────
    # {
    #     "name":       "NewPlatform",
    #     "dept":       "部门名",
    #     "channel":    "推广方式",
    #     "excel_dir":  r"D:\NewPlatform自动播报",
    #     "excel_glob": "日报-大盘日报_*.xlsx",
    #     "output_dir": r"D:\自动播报报告",
    #     "bot_token":  "<Token>",
    #     "chat_id":    "<ChatID>",
    #     "n_days":     35,
    # },
]

SEND_TO_TG = True   # ★ 全局开关，False 时只生成 PDF 不发送

# ══════════════════════════════════════════════════════
# 全局常量
# ══════════════════════════════════════════════════════
TODAY = pd.Timestamp(date.today())
DASH  = "—"

FONT_WIN = [
    r"C:\Windows\Fonts\msyh.ttc",
    r"C:\Windows\Fonts\simsun.ttc",
    r"C:\Windows\Fonts\simhei.ttf",
]

# ── 颜色分区（按列索引，0-based）─────────────────────
BLUE_START   = 36   # 首充次日复登率
BLUE_END     = 48   # 首充7日复投率（含）
ORANGE_START = 49   # 老用户付费率
ORANGE_END   = 60   # 总赠送充值比（含）

C_BLUE_HDR   = colors.HexColor("#1F4E79")
C_ORANGE_HDR = colors.HexColor("#833C00")

# ── 留存/LTV 完整性窗口 ───────────────────────────────
RETENTION_N = {
    "首充2日复登率":  2,  "首充3日复登率":  3,  "首充7日复登率":  7,
    "首充14日复登率": 14, "首充30日复登率": 30,
    "首充2日复充率":  2,  "首充3日复充率":  3,  "首充7日复充率":  7,
    "首充14日复充率": 14, "首充30日复充率": 30,
    "首充2日复投率":  2,  "首充3日复投率":  3,  "首充7日复投率":  7,
}
LTV_N = {"首充LTV7日": 7, "首充LTV14日": 14, "首充LTV30日": 30}

# ══════════════════════════════════════════════════════
# 字体（全局注册一次）
# ══════════════════════════════════════════════════════
_FN  = "STSong-Light"
_FNB = "STSong-Light"
_FONT_REGISTERED = False


def register_fonts():
    global _FN, _FNB, _FONT_REGISTERED
    if _FONT_REGISTERED:
        return
    from reportlab.lib.fonts import addMapping

    def try_ttf(path):
        for kw in [{"subfontIndex": 0}, {}]:
            try:
                pdfmetrics.registerFont(TTFont("CH",   path, **kw))
                pdfmetrics.registerFont(TTFont("CH_B", path, **kw))
                return True
            except Exception:
                continue
        return False

    for path in FONT_WIN:
        if os.path.exists(path) and try_ttf(path):
            addMapping("CH", 0, 0, "CH")
            addMapping("CH", 1, 0, "CH_B")
            addMapping("CH", 0, 1, "CH")
            addMapping("CH", 1, 1, "CH_B")
            _FN, _FNB = "CH", "CH_B"
            print(f"🔤 字体: {path}")
            _FONT_REGISTERED = True
            return

    pdfmetrics.registerFont(UnicodeCIDFont("STSong-Light"))
    _FN = _FNB = "STSong-Light"
    print("🔤 字体: STSong-Light (CID)")
    _FONT_REGISTERED = True


# ══════════════════════════════════════════════════════
# 数据读取
# ══════════════════════════════════════════════════════
def clean_num(s):
    return pd.to_numeric(
        s.astype(str).str.replace(",", "").str.replace("%", ""),
        errors="coerce"
    )


def find_excel(cfg) -> str:
    pattern = os.path.join(cfg["excel_dir"], cfg["excel_glob"])
    files = sorted(glob.glob(pattern))
    if files:
        return files[-1]
    raise FileNotFoundError(f"[{cfg['name']}] 未找到: {pattern}")


def load_data(filepath, n_days) -> pd.DataFrame:
    df = pd.read_excel(filepath)
    df["日期"] = pd.to_datetime(df["日期"].astype(str), format="%Y%m%d", errors="coerce")
    for col in df.columns[1:]:
        df[col] = clean_num(df[col])
    df = df.dropna(subset=["日期"])
    df = df.sort_values("日期", ascending=False).head(n_days)
    df = df.sort_values("日期", ascending=True).reset_index(drop=True)
    return df


# ══════════════════════════════════════════════════════
# 工具函数
# ══════════════════════════════════════════════════════
def is_complete(col, row_date):
    n = RETENTION_N.get(col) or LTV_N.get(col)
    if n is None:
        return True
    return (TODAY - row_date).days >= n

def fdate(v):
    if isinstance(v, str): return v
    try: return v.strftime("%m/%d")
    except: return str(v)

def fint(v):
    if v is None or (isinstance(v, float) and (np.isnan(v) or np.isinf(v))): return DASH
    return f"{int(round(v)):,}"

def f2(v):
    if v is None or (isinstance(v, float) and (np.isnan(v) or np.isinf(v))): return DASH
    return f"{v:.2f}"

def fpct(v):
    if v is None or (isinstance(v, float) and (np.isnan(v) or np.isinf(v))): return DASH
    return f"{v:.2f}%"


# ══════════════════════════════════════════════════════
# 列定义（61列）
# ══════════════════════════════════════════════════════
def build_cols(platform, dept, channel):
    COLS = [
        # ── 基础信息（索引0~3）──
        ("_plat",  "平台",       str,   "str",    platform),
        ("_dept",  "推广\n部门", str,   "str",    dept),
        ("_chan",  "推广\n方式", str,   "str",    channel),
        ("日期",   "日期",       fdate, "date",   None),
        # ── 消耗/成本/LTV（索引4~9）──
        ("真实消耗",      "总消耗\n(U)",      fint, "sum_int", None),
        ("_fc_cost",      "首充\n成本(U)",    f2,   "calc_fc", None),
        ("_l1_cost",      "一级首充\n成本(U)", f2,  "calc_l1", None),
        ("首充LTV7日",    "7日\nLTV",         f2,   "ltv",     "首充人数"),
        ("首充LTV14日",   "15日\nLTV",        f2,   "ltv",     "首充人数"),
        ("首充LTV30日",   "30日\nLTV",        f2,   "ltv",     "首充人数"),
        # ── 充提差（索引10~11）──
        ("充提差",        "总充提差\n(U)",    fint, "sum_int", None),
        ("充提差比",      "充提差\n比",       fpct, "wpct",    ("充提差", "充值金额")),
        # ── 历史（索引12~14）──
        ("历史累计毛利",      "历史营收\n(历史充提差\n-历史消耗)", fint, "last", None),
        ("历史累计充提差",    "历史充提差",   fint, "last",    None),
        ("历史累计真实消耗",  "历史消耗",     fint, "last",    None),
        # ── 充值基础（索引15~18）──
        ("充值金额",      "充值金额\n(U)",    fint, "sum_int", None),
        ("充值人数",      "充值人数",         fint, "sum_int", None),
        ("人均充值金额",  "人均充值\n金额(U)", f2,  "wdiv",    ("充值金额", "充值人数")),
        ("人均充值笔数",  "人均充值\n笔数",   f2,   "wdiv",    ("_total_orders", "充值人数")),
        # ── 首充/新充（索引19~23）──
        ("首充人数",      "首充人数",         fint, "sum_int", None),
        ("新充人数",      "新充人数",         fint, "sum_int", None),
        ("首充人均充值金额", "首充人均\n充值(U)", f2, "wdiv",  ("_fc_amt", "首充人数")),
        ("_old_arppu",    "老用户人均\n充值(U)", f2, "wdiv",  ("老用户充值金额", "_old_users")),
        ("首充人均充值笔数", "首充人均\n充值笔",  f2, "wdiv",  ("_fc_orders", "首充人数")),
        # ── 一级/非一级（索引24~30）──
        ("一级首充人数",  "一级首充\n人数",   fint, "sum_int", None),
        ("一级新充人数",  "一级新充\n人数",   fint, "sum_int", None),
        ("非一级充值人数", "非一级充值\n人数", fint, "sum_int", None),
        ("非一级首充人数/充值人数", "非一级首充\n人数/充值\n人数", fpct, "wpct", ("非一级首充人数", "充值人数")),
        ("非一级充值人数/充值人数", "非一级充值\n人数/充值\n人数", fpct, "wpct", ("非一级充值人数", "充值人数")),
        ("非一级首充人数/首充人数", "非一级首充\n人数/首充\n人数", fpct, "wpct", ("非一级首充人数", "首充人数")),
        ("_inv_rate",     "邀请首充\n人数/首充\n人数", fpct, "dash", None),
        # ── 赢钱（索引31~32）──
        ("_cz_win",       "充值赢钱\n人数占比", fpct, "czwin", None),
        ("_sc_win",       "首充赢钱\n人数占比", fpct, "scwin", None),
        # ── 注册/转化（索引33~35）──
        ("注册人数",      "新增注册\n人数",   fint, "sum_int", None),
        ("新充付费率",    "新增充值\n转化率", fpct, "wpct",    ("新充人数", "注册人数")),
        ("首充转化率",    "首充转化\n率",     fpct, "wpct",    ("首充人数", "注册人数")),
        # ══ 蓝色留存区（索引36~48）══
        ("首充2日复登率",  "首充次日\n复登率",    fpct, "ret",  "首充人数"),
        ("首充3日复登率",  "首充3日\n复登率",     fpct, "ret",  "首充人数"),
        ("首充7日复登率",  "首充7日\n复登率",     fpct, "ret",  "首充人数"),
        ("首充2日复充率",  "首充次日\n复充率",    fpct, "ret",  "首充人数"),
        ("首充3日复充率",  "首充3日\n复充率",     fpct, "ret",  "首充人数"),
        ("首充7日复充率",  "首充7日\n复充率",     fpct, "ret",  "首充人数"),
        ("首充30日复充率", "30日首充\n用户复充率", fpct, "ret", "首充人数"),
        ("_dep2r",         "充值次日\n复充率",    fpct, "dash", None),
        ("_dep3r",         "充值3日\n复充率",     fpct, "dash", None),
        ("_dep7r",         "充值7日\n复充率",     fpct, "dash", None),
        ("首充2日复投率",  "首充次日\n复投率",    fpct, "ret",  "首充人数"),
        ("首充3日复投率",  "首充3日\n复投率",     fpct, "ret",  "首充人数"),
        ("首充7日复投率",  "首充7日\n复投率",     fpct, "ret",  "首充人数"),
        # ══ 橙色区（索引49~60）══
        # 出入比 = Σ提款金额 / Σ充值金额，提款金额在 augment 中由原始出入比反推
        ("老充付费率",    "老用户付\n费率",       fpct, "wavg",    "充值人数"),
        ("老用户充值金额", "老用户充值\n金额(U)", fint, "sum_int", None),
        ("出入比",        "出入比",               fpct, "wpct",    ("_withdrawal", "充值金额")),
        # ★ 修改2：游戏盈亏投注比表头改为"盈亏率"
        ("游戏盈亏投注比", "盈亏率",              fpct, "wavg",    "投注金额"),  # ★ 修改2
        ("投注金额",      "投注金额\n(U)",        fint, "sum_int", None),
        ("投充比",        "投充比",               fpct, "wpct",    ("投注金额", "充值金额")),
        ("投注人数",      "投注用户\n数",         fint, "sum_int", None),
        ("充投人数",      "充投用户\n数",         fint, "sum_int", None),
        ("获得佣金人数",  "佣金人数",             fint, "sum_int", None),
        ("获得佣金金额",  "佣金金额\n(U)",        fint, "sum_int", None),
        ("_inv_fc",       "代理邀请\n首充人数",   fpct, "dash",    None),
        ("总赠送充值比",  "总赠送\n充值比",       fpct, "wavg",    "充值金额"),
    ]
    assert len(COLS) == 61, f"COLS长度={len(COLS)}，期望61"
    return COLS


# ══════════════════════════════════════════════════════
# augment：增加辅助计算列
# ══════════════════════════════════════════════════════
def augment(df, platform, dept, channel):
    d = df.copy()
    d["_plat"] = platform
    d["_dept"] = dept
    d["_chan"]  = channel

    d["_old_users"] = (d["充值人数"] - d["首充人数"]).clip(lower=0)
    d["_old_arppu"]  = np.where(d["_old_users"] > 0,
                                 d["老用户充值金额"] / d["_old_users"], np.nan)
    d["_fc_cost"]    = np.where(d["首充人数"] > 0,
                                 d["真实消耗"] / d["首充人数"], np.nan)
    d["_l1_cost"]    = np.where(d["一级首充人数"] > 0,
                                 d["真实消耗"] / d["一级首充人数"], np.nan)

    d["_cz_win"] = np.where(
        d["充值人数"] > 0,
        (d["平台首充用户赢钱比例"] / 100 * d["首充人数"] +
         d["平台非首充用户赢钱比例"] / 100 * d["_old_users"]) / d["充值人数"] * 100,
        np.nan
    )
    d["_sc_win"] = d["平台首充用户赢钱比例"]

    # LTV 不完整行置 NaN
    for col, n in LTV_N.items():
        if col in d.columns:
            d[col] = np.where(
                d["日期"].apply(lambda dt: (TODAY - dt).days >= n),
                d[col], np.nan
            )

    d["_total_orders"] = d["人均充值笔数"]     * d["充值人数"]
    d["_fc_amt"]       = d["首充人均充值金额"]  * d["首充人数"]
    d["_fc_orders"]    = d["首充人均充值笔数"]  * d["首充人数"]

    # 出入比 = 提款金额 / 充值金额；Excel 存的是百分数（如79.67），反推提款金额供 wpct 使用
    d["投充比"]     = np.where(d["充值金额"] > 0, d["投注金额"] / d["充值金额"] * 100, np.nan)
    d["_withdrawal"] = np.where(d["充值金额"] > 0, d["出入比"] / 100 * d["充值金额"], np.nan)
    # 占位列（xlsx 无此字段，显示 —）
    d["_inv_rate"] = np.nan
    d["_dep2r"]    = np.nan
    d["_dep3r"]    = np.nan
    d["_dep7r"]    = np.nan
    d["_inv_fc"]   = np.nan

    return d


# ══════════════════════════════════════════════════════
# 合计/日均计算
# ══════════════════════════════════════════════════════
def calc_summary(aug, COLS, is_avg=False):
    n = len(aug)
    S = {}

    def wpct(nc, dc):
        ns = aug[nc].sum() if nc in aug.columns else 0
        ds = aug[dc].sum() if dc in aug.columns else 0
        return ns / ds * 100 if ds else np.nan

    def wdiv(nc, dc):
        ns = aug[nc].sum() if nc in aug.columns else 0
        ds = aug[dc].sum() if dc in aug.columns else 0
        return ns / ds if ds else np.nan

    def wavg(col, wcol):
        if col not in aug.columns or wcol not in aug.columns: return np.nan
        v, w = aug[col], aug[wcol]
        ok = v.notna() & w.notna() & (w > 0)
        return (v[ok] * w[ok]).sum() / w[ok].sum() if ok.any() else np.nan

    def wret(col, wcol):
        if col not in aug.columns or wcol not in aug.columns: return np.nan
        complete = aug["日期"].apply(lambda d: is_complete(col, d))
        v, w = aug[col], aug[wcol]
        ok = complete & v.notna() & w.notna() & (w > 0)
        return (v[ok] * w[ok]).sum() / w[ok].sum() if ok.any() else np.nan

    def wltv(col, wcol):
        if col not in aug.columns or wcol not in aug.columns: return np.nan
        v, w = aug[col], aug[wcol]
        ok = v.notna() & w.notna() & (w > 0)
        return (v[ok] * w[ok]).sum() / w[ok].sum() if ok.any() else np.nan

    for key, hdr, fmt, stype, extra in COLS:
        if stype == "str":
            S[key] = extra
        elif stype == "date":
            S[key] = "日均" if is_avg else "合计"
        elif stype == "last":
            S[key] = aug[key].iloc[-1] if key in aug.columns else np.nan
        elif stype == "sum_int":
            v = aug[key].sum() if key in aug.columns else np.nan
            S[key] = v / n if is_avg else v
        elif stype == "wpct":
            S[key] = wpct(extra[0], extra[1])
        elif stype == "wdiv":
            S[key] = wdiv(extra[0], extra[1])
        elif stype == "wavg":
            # ★ 修改4：出入比/盈亏率/老充付费率等均走 wavg，直接对原列加权平均
            S[key] = wavg(key, extra)
        elif stype == "ret":
            S[key] = wret(key, extra)
        elif stype == "ltv":
            S[key] = wltv(key, extra)
        elif stype == "czwin":
            fc_w  = (aug["平台首充用户赢钱比例"] / 100 * aug["首充人数"]).sum()
            old_w = (aug["平台非首充用户赢钱比例"] / 100 * aug["_old_users"]).sum()
            tot   = aug["充值人数"].sum()
            S[key] = (fc_w + old_w) / tot * 100 if tot else np.nan
        elif stype == "scwin":
            S[key] = wavg("_sc_win", "首充人数")
        elif stype == "calc_fc":
            d = aug["首充人数"].sum()
            S[key] = aug["真实消耗"].sum() / d if d else np.nan
        elif stype == "calc_l1":
            d = aug["一级首充人数"].sum()
            S[key] = aug["真实消耗"].sum() / d if d else np.nan
        elif stype == "dash":
            S[key] = np.nan
        else:
            S[key] = np.nan
    return S


# ══════════════════════════════════════════════════════
# 单行格式化
# ══════════════════════════════════════════════════════
def fmt_row(aug_row, COLS, is_sum=False):
    cells = []
    for key, hdr, fmt, stype, extra in COLS:
        if stype == "str":
            v = aug_row.get(key, extra) if isinstance(aug_row, dict) else extra
            cells.append(str(v)); continue
        if stype == "dash":
            cells.append(DASH); continue

        if is_sum:
            v = aug_row.get(key, np.nan)
            cells.append(DASH if (v is None or (isinstance(v, float) and (np.isnan(v) or np.isinf(v))))
                         else fmt(v))
            continue

        row_date = aug_row["日期"]
        if stype in ("ret", "ltv") or key in RETENTION_N or key in LTV_N:
            if not is_complete(key, row_date):
                cells.append(DASH); continue

        v = aug_row[key] if key in aug_row.index else np.nan
        cells.append(DASH if (v is None or (isinstance(v, float) and (np.isnan(v) or np.isinf(v))))
                     else fmt(v))
    return cells


# ══════════════════════════════════════════════════════
# 生成 PDF
# ══════════════════════════════════════════════════════
def generate_pdf(df, output_path, cfg):
    platform = cfg["name"]
    dept     = cfg.get("dept", "")
    channel  = cfg.get("channel", "")
    n_days   = cfg.get("n_days", 35)

    COLS = build_cols(platform, dept, channel)
    aug  = augment(df, platform, dept, channel)

    sum_d = calc_summary(aug, COLS, is_avg=False)
    avg_d = calc_summary(aug, COLS, is_avg=True)

    headers    = [c[1] for c in COLS]
    sum_row    = fmt_row(sum_d, COLS, is_sum=True)
    avg_row    = fmt_row(avg_d, COLS, is_sum=True)
    data_rows  = [fmt_row(r, COLS) for _, r in aug.iterrows()]
    table_data = [headers, sum_row, avg_row] + data_rows

    # ── 列宽（mm）────────────────────────────────────
    W = {
        "_plat":9, "_dept":9, "_chan":9, "日期":10,
        "真实消耗":14, "_fc_cost":11, "_l1_cost":12,
        "首充LTV7日":9, "首充LTV14日":9, "首充LTV30日":9,
        "充提差":14, "充提差比":9,
        "历史累计毛利":16, "历史累计充提差":13, "历史累计真实消耗":13,
        "充值金额":14, "充值人数":9,
        "人均充值金额":10, "人均充值笔数":9,
        "首充人数":9, "新充人数":9,
        "首充人均充值金额":10, "_old_arppu":10, "首充人均充值笔数":9,
        "一级首充人数":9, "一级新充人数":9, "非一级充值人数":9,
        "非一级首充人数/充值人数":10, "非一级充值人数/充值人数":10,
        "非一级首充人数/首充人数":10, "_inv_rate":10,
        "_cz_win":9, "_sc_win":9,
        "注册人数":9, "新充付费率":9, "首充转化率":9,
        "首充2日复登率":9, "首充3日复登率":9, "首充7日复登率":9,
        "首充2日复充率":9, "首充3日复充率":9, "首充7日复充率":9,
        "首充30日复充率":10,
        "_dep2r":9, "_dep3r":9, "_dep7r":9,
        "首充2日复投率":9, "首充3日复投率":9, "首充7日复投率":9,
        "老充付费率":9, "老用户充值金额":12,
        "出入比":9, "游戏盈亏投注比":8,
        "投注金额":14, "投充比":9,
        "投注人数":9, "充投人数":9,
        "获得佣金人数":9, "获得佣金金额":11,
        "_inv_fc":9, "总赠送充值比":9,
    }
    col_widths = [W.get(c[0], 9) * mm for c in COLS]

    PAGE_H = 297 * mm
    MARGIN = 5  * mm
    PAGE_W = sum(col_widths) + 2 * MARGIN

    HDR_H  = 15 * mm
    SUM_H  = 7  * mm
    DATA_H = 6  * mm
    row_heights = [HDR_H, SUM_H, SUM_H] + [DATA_H] * len(data_rows)

    doc = SimpleDocTemplate(
        output_path,
        pagesize=(PAGE_W, PAGE_H),
        leftMargin=MARGIN, rightMargin=MARGIN,
        topMargin=4 * mm, bottomMargin=4 * mm,
    )

    ts = ParagraphStyle("T", fontName=_FN, fontSize=11, leading=15, alignment=1, spaceAfter=2)
    ss = ParagraphStyle("S", fontName=_FN, fontSize=7,  leading=10, alignment=1, spaceAfter=2)

    d0 = df["日期"].iloc[0].strftime("%Y-%m-%d")
    d1 = df["日期"].iloc[-1].strftime("%Y-%m-%d")
    dept_str = f"　推广部门: {dept}" if dept else ""
    chan_str  = f"　推广方式: {channel}" if channel else ""
    story = [
        Paragraph(f"{platform} 大盘日报 近{n_days}天数据（{d0} ~ {d1}）", ts),
        Paragraph(f"平台: {platform}{dept_str}{chan_str}　生成时间: {date.today()}", ss),
        Spacer(1, 2 * mm),
    ]

    # ── 颜色常量 ──────────────────────────────────────
    C_HDR_MAIN = colors.HexColor("#1F3864")
    C_HDR_FONT = colors.white
    C_SUM_BG   = colors.HexColor("#FFF2CC")
    C_AVG_BG   = colors.HexColor("#E2EFDA")
    C_ALT_MAIN = colors.HexColor("#EBF3FB")
    C_GRID     = colors.HexColor("#BBBBBB")

    n_rows = len(table_data)

    cmds = [
        ("FONTNAME",      (0, 0), (-1, 0),  _FN),
        ("FONTSIZE",      (0, 0), (-1, 0),  5.5),
        ("FONTNAME",      (0, 1), (-1, -1), _FN),
        ("FONTSIZE",      (0, 1), (-1, -1), 5.2),
        ("FONTNAME",      (0, 1), (-1, 2),  _FNB),
        ("FONTSIZE",      (0, 1), (-1, 2),  5.5),
        ("ALIGN",         (0, 0), (-1, -1), "CENTER"),
        ("VALIGN",        (0, 0), (-1, -1), "MIDDLE"),
        ("GRID",          (0, 0), (-1, -1), 0.3, C_GRID),
        ("LINEBELOW",     (0, 0), (-1, 0),  0.8, colors.white),
        ("TOPPADDING",    (0, 0), (-1, -1), 1),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 1),
        ("LEFTPADDING",   (0, 0), (-1, -1), 1),
        ("RIGHTPADDING",  (0, 0), (-1, -1), 1),
        ("BACKGROUND",    (0, 1), (-1, 1),  C_SUM_BG),
        ("BACKGROUND",    (0, 2), (-1, 2),  C_AVG_BG),
    ]

    # 表头分区着色
    cmds.append(("BACKGROUND", (0, 0),            (BLUE_START - 1, 0), C_HDR_MAIN))
    cmds.append(("TEXTCOLOR",  (0, 0),            (BLUE_START - 1, 0), C_HDR_FONT))
    cmds.append(("BACKGROUND", (BLUE_START, 0),   (BLUE_END, 0),       C_BLUE_HDR))
    cmds.append(("TEXTCOLOR",  (BLUE_START, 0),   (BLUE_END, 0),       C_HDR_FONT))
    cmds.append(("BACKGROUND", (ORANGE_START, 0), (ORANGE_END, 0),     C_ORANGE_HDR))
    cmds.append(("TEXTCOLOR",  (ORANGE_START, 0), (ORANGE_END, 0),     C_HDR_FONT))

    # 数据行交替背景
    for ri in range(3, n_rows):
        if ri % 2 == 0:
            cmds.append(("BACKGROUND", (0, ri), (-1, ri), C_ALT_MAIN))

    tbl = Table(table_data, colWidths=col_widths, rowHeights=row_heights, repeatRows=1)
    tbl.setStyle(TableStyle(cmds))
    story.append(tbl)
    doc.build(story)
    print(f"  ✅ PDF 已生成：{output_path}")


# ══════════════════════════════════════════════════════
# Telegram 发送
# ══════════════════════════════════════════════════════
def send_telegram(pdf_path, cfg):
    url = f"https://api.telegram.org/bot{cfg['bot_token']}/sendDocument"
    with open(pdf_path, "rb") as f:
        resp = requests.post(
            url,
            data={"chat_id": cfg["chat_id"], "parse_mode": "Markdown"},
            files={"document": (os.path.basename(pdf_path), f, "application/pdf")},
            timeout=60
        )
    ok = resp.status_code == 200
    print(f"  {'✅' if ok else '❌'} TG {'成功' if ok else resp.text}")
    return ok


# ══════════════════════════════════════════════════════
# 单平台处理
# ══════════════════════════════════════════════════════
def run_platform(cfg):
    name = cfg["name"]
    print(f"\n{'=' * 50}")
    print(f"  平台：{name}")
    print(f"{'=' * 50}")
    try:
        filepath = find_excel(cfg)
        print(f"  📂 读取: {os.path.basename(filepath)}")

        df = load_data(filepath, cfg.get("n_days", 35))
        print(f"  ✅ {len(df)} 天: {df['日期'].iloc[0].date()} ~ {df['日期'].iloc[-1].date()}")

        output_dir = cfg.get("output_dir", os.path.dirname(os.path.abspath(filepath)))
        Path(output_dir).mkdir(parents=True, exist_ok=True)

        today_str = date.today().strftime("%Y%m%d")
        pdf_path  = os.path.join(output_dir, f"{name}-大盘报表-{today_str}.pdf")

        generate_pdf(df, pdf_path, cfg)

        if SEND_TO_TG:
            print(f"  📤 发送 TG...")
            send_telegram(pdf_path, cfg)

    except FileNotFoundError as e:
        print(f"  ⚠️  跳过（找不到文件）: {e}")
    except Exception as e:
        import traceback
        print(f"  ❌ 失败: {e}")
        traceback.print_exc()


# ══════════════════════════════════════════════════════
# 主流程
# ══════════════════════════════════════════════════════
def main():
    register_fonts()
    print(f"\n{'=' * 50}")
    print(f"  多平台大盘日报  {date.today()}")
    print(f"  共 {len(PLATFORMS)} 个平台")
    print(f"{'=' * 50}")

    for cfg in PLATFORMS:
        run_platform(cfg)

    print(f"\n🎉 全部完成！")


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
"""
多平台大盘日报生成器 v6  (PDF + Telegram)
=========================================
v6 改动（基于 v5）：
  1. 字段重命名：「首充获客成本」→「一级首充获客成本」
     - load_df、chart_cac_bonus、build_pdf（KPI 显示、均值）全部对齐
  2. 自适应数据量：去掉所有隐式 60 天假设
     - 图表 x 轴刻度、留存窗口截止、KPI 均值窗口均根据实际行数动态计算
     - 数据行数 < 留存 lag 时自动降级处理，不报错
     - _xticks 步长自适应（保证标签不超过 20 个）
     - KPI 汇总中「近 N 日」N 取实际有效天数
  3. 其余逻辑与 v5 完全一致
"""

import subprocess, sys

def _ensure(pkg, import_name=None):
    name = import_name or pkg
    try: __import__(name)
    except ImportError:
        for extra in [[], ["--break-system-packages"]]:
            if subprocess.call([sys.executable, "-m", "pip", "install", pkg, "-q"] + extra,
                               stderr=subprocess.DEVNULL) == 0:
                break

_ensure("reportlab"); _ensure("matplotlib"); _ensure("fonttools")

import pandas as pd
import numpy as np
import requests, io, logging, os, tempfile, platform as _platform
from pathlib import Path
from datetime import datetime

import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import matplotlib.font_manager as fm
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

from reportlab.lib.pagesizes import A4
from reportlab.lib.units import mm
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.platypus import (SimpleDocTemplate, Paragraph, Spacer,
                                Table, TableStyle, Image,
                                HRFlowable, KeepTogether)
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont

# ══════════════════════════════════════════════════════════════════════════════
# ★ 多平台配置
# ══════════════════════════════════════════════════════════════════════════════
PLATFORMS = [

    {
        "name":       "Afun_mx",
        "excel_dir":  r"D:\Afun_mx自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
    },
    {
        "name":       "Soluno",
        "excel_dir":  r"D:\soluno自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",

        # 下面这个是会发到工作群的，先在上面测试成功再发
    },
    {
        "name":       "sol777",
        "excel_dir":  r"D:\sol777自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
    },
    {
        "name":       "Lucro",
        "excel_dir":  r"D:\Lucro自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
    },
    {
        "name":       "3bet",
        "excel_dir":  r"D:\3bet自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
    },
    {
        "name":       "A7X",
        "excel_dir":  r"D:\A7X自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
    },
    {
        "name":       "Wey7",
        "excel_dir":  r"D:\wey7自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
                # 下面这个是会发到工作群的，先在上面测试成功再发
    },
    {
        "name":       "oro7x",
        "excel_dir":  r"D:\oro7x自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
    },
    {
        "name":       "Letjoy",
        "excel_dir":  r"D:\Letjoy自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
    },
    {
        "name":       "G777",
        "dept":       "",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\G777自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },
    {
        "name":       "oro7x",
        "dept":       "",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\G777自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },
    {
        "name":       "Luna777",
        "dept":       "",
        "channel":    "短信/投放",
        "excel_dir":  r"D:\Luna777自动播报",
        "excel_glob": "日报-大盘日报_*.xlsx",
        "output_dir": r"D:\自动播报报告",
        "bot_token":  "8782270942:AAFng-sm_nckrJpdgrinkcx-77iYOYuDy2k",
        "chat_id":    "-5258799427",
        "n_days":     35,
    },

    # 新增盘：复制下方模板并修改
    # {
    #     "name":       "Afun_br",
    #     "excel_dir":  r"D:\Afun_br自动播报",
    #     "excel_glob": "日报-大盘日报_*.xlsx",
    #     "output_dir": r"D:\Afun_br自动播报\reports",
    #     "bot_token":  "<BR盘 Token>",
    #     "chat_id":    "<BR盘 ChatID>",
    # },


]

# ★ 留存目标（lag = 窗口天数-1）
RETENTION_TARGETS = [
    ("2日复充",  "首充2日复充率",  21.0,  1),
    ("3日复充",  "首充3日复充率",  15.0,  2),
    ("7日复充",  "首充7日复充率",  11.0,  6),
    ("14日复充", "首充14日复充率",  8.0, 13),
    ("30日复充", "首充30日复充率",  6.0, 29),
]

# ══════════════════════════════════════════════════════════════════════════════
# 字体初始化
# ══════════════════════════════════════════════════════════════════════════════
logging.basicConfig(level=logging.INFO, format="%(asctime)s  %(message)s",
                    datefmt="%H:%M:%S", handlers=[logging.StreamHandler(sys.stdout)])
log = logging.getLogger(__name__)


def _init_font() -> str:
    from fontTools.ttLib import TTFont as FTF, TTCollection
    cache = os.path.join(tempfile.gettempdir(), "cjk_rl_font.ttf")
    if os.path.exists(cache):
        try:
            if "glyf" in FTF(cache): return cache
        except Exception: pass
    sys_name = _platform.system()
    candidates = {
        "Windows": [os.path.join(os.environ.get("WINDIR", r"C:\Windows"), "Fonts", n)
                    for n in ["msyh.ttc", "simhei.ttf", "simsun.ttc", "msyhbd.ttc"]],
        "Darwin":  ["/System/Library/Fonts/PingFang.ttc",
                    "/Library/Fonts/Arial Unicode.ttf"],
    }.get(sys_name, [
        "/usr/share/fonts/truetype/wqy/wqy-microhei.ttc",
        "/usr/share/fonts/truetype/wqy/wqy-zenhei.ttc",
        "/usr/share/fonts/opentype/noto/NotoSansCJK-Regular.ttc",
    ])
    def _try(path):
        try:
            ext = os.path.splitext(path)[1].lower()
            if ext == ".ttf":
                f = FTF(path)
                if "glyf" in f: return path
            else:
                for f in TTCollection(path):
                    if "glyf" in f: f.save(cache); return cache
        except Exception: pass
    for p in candidates:
        if os.path.exists(p):
            r = _try(p)
            if r: return r
    import urllib.request
    log.info("下载备用字体 WQY MicroHei …")
    wqy = os.path.join(tempfile.gettempdir(), "wqy-microhei.ttc")
    urllib.request.urlretrieve(
        "https://github.com/anthonyfok/fonts-wqy-microhei/raw/master/wqy-microhei.ttc", wqy)
    r = _try(wqy)
    if r: return r
    raise RuntimeError("无法找到可用的 TrueType 中文字体。")


_FONT_PATH = _init_font()
pdfmetrics.registerFont(TTFont("NotoSC",   _FONT_PATH))
pdfmetrics.registerFont(TTFont("NotoSC-B", _FONT_PATH))
_FP = fm.FontProperties(fname=_FONT_PATH)
plt.rcParams["axes.unicode_minus"] = False

# ══════════════════════════════════════════════════════════════════════════════
# 调色板
# ══════════════════════════════════════════════════════════════════════════════
C_GOLD   = "#B5922A"; C_DARK  = "#2C2416"; C_RED    = "#C04A3A"
C_GREEN  = "#3D7A5E"; C_BLUE  = "#4A7AB5"; C_MUTED  = "#8A7A68"
C_ORANGE = "#D4743A"; C_GRAY  = "#AAAAAA"

# ══════════════════════════════════════════════════════════════════════════════
# ★ v6 新增：获客成本字段名常量（统一修改入口）
# ══════════════════════════════════════════════════════════════════════════════
# ★ 获客成本列名：若 Excel 里不是这个名字，程序会在 load_df 中自动搜索含「获客成本」的列
# 常见别名：「首充获客成本」「一级首充获客成本」「一级首充获客成本$」等
# 只需把 Excel 里实际使用的列名填在这里即可
CAC_COL = "一级首充获客成本"

# ══════════════════════════════════════════════════════════════════════════════
# 1. 数据加载
# ══════════════════════════════════════════════════════════════════════════════

def find_excel(cfg) -> Path:
    files = sorted(Path(cfg["excel_dir"]).glob(cfg["excel_glob"]),
                   key=lambda f: f.stat().st_mtime, reverse=True)
    if not files:
        raise FileNotFoundError(f"[{cfg['name']}] 未找到 {cfg['excel_glob']}")
    log.info(f"[{cfg['name']}] 读取：{files[0].name}")
    return files[0]


def clean_num(s):
    return pd.to_numeric(
        s.astype(str).str.replace(",", "", regex=False).str.replace("%", "", regex=False),
        errors="coerce")


def load_df(path: Path) -> pd.DataFrame:
    df = pd.read_excel(path)
    for col in df.columns[1:]:
        df[col] = clean_num(df[col])

    # 统一转为旧→新（自动兼容升降序）
    df = df.sort_values("日期", ascending=True).reset_index(drop=True)

    df.rename(columns={"游戏盈亏投注比": "游戏盈利率"}, inplace=True)

    # ★ 诊断：打印所有列名，方便排查字段对不上的问题
    log.info(f"[load_df] Excel 列名列表：{df.columns.tolist()}")

    # ★ 模糊匹配兜底：若 CAC_COL 不在列中，尝试找包含「获客成本」的列并重命名
    if CAC_COL not in df.columns:
        candidates = [c for c in df.columns if "获客成本" in str(c)]
        if candidates:
            best = candidates[0]
            log.warning(f"[load_df] 未找到列「{CAC_COL}」，自动使用「{best}」代替（完整候选：{candidates}）")
            df.rename(columns={best: CAC_COL}, inplace=True)
        else:
            log.warning(f"[load_df] 未找到任何含「获客成本」的列，获客成本将显示 N/A。"
                        f"请检查 Excel 列名是否为：{CAC_COL}")
    df["日期标签"] = df["日期"].astype(str).apply(lambda s: f"{s[4:6]}/{s[6:8]}")

    df["老用户ARPPU"] = (df["老用户充值金额"] /
        (df["充值人数"] - df["首充人数"]).replace(0, np.nan)).round(3)
    df["累计毛利ROI"] = (df["历史累计毛利"] /
        df["历史累计真实消耗"].replace(0, np.nan) * 100).round(2)
    # 日充提差ROI = 充提差 / 真实消耗
    df["日充提差ROI"] = (
        df["充提差"] /
        df["真实消耗"].replace(0, np.nan) * 100
    ).round(2)

    # 充值=0 时置 NaN
    zero_dep = df["充值金额"].fillna(0) <= 0
    for c in ["充提差", "充提差比", "充值人数", "人均充值金额", "首充人数",
              "老用户充值金额", "投注金额", "新充付费率", "老充付费率",
              "首充转化率", "总赠送充值比", "游戏盈利率",
              "首充2日活跃留存率", "首充2日复充率", "首充3日活跃留存率",
              "首充7日活跃留存率", "首充3日复充率", "首充7日复充率",
              "首充14日复充率", "首充30日复充率",
              "首充LTV7日", "首充LTV14日", "首充LTV30日"]:
        if c in df.columns:
            df.loc[zero_dep, c] = np.nan

    # ★ 最新一天真实消耗和一级首充获客成本置NaN（当日录入未完整）
    df.loc[df.index[-1], "真实消耗"] = np.nan
    if CAC_COL in df.columns:
        df.loc[df.index[-1], CAC_COL] = np.nan

    # ★ 关键：zero_cost 必须在最新一天置NaN之后再判断，否则会误把最新一天之前的
    #   正常行（恰好消耗字段值为0）判定为无效，导致获客成本均值异常显示N/A
    zero_cost = df["真实消耗"].fillna(0) <= 0
    if CAC_COL in df.columns:
        df.loc[zero_cost, CAC_COL] = np.nan

    # 首充 < 50 时留存置NaN
    small = df["首充人数"].fillna(0) < 50
    for c in ["首充2日活跃留存率", "首充2日复充率", "首充3日活跃留存率",
              "首充7日活跃留存率", "首充3日复充率", "首充7日复充率",
              "首充14日复充率", "首充30日复充率"]:
        if c in df.columns:
            df.loc[small, c] = np.nan
            df[c] = df[c].clip(0, 100)
    return df


# ══════════════════════════════════════════════════════════════════════════════
# 2. 留存工具
# ══════════════════════════════════════════════════════════════════════════════

def get_complete_series(df, col, lag):
    """窗口已关闭且值>0的全量数据 (xs, vals, dates)
    ★ v6：cutoff 基于实际行数，不足 lag+1 行则返回空
    """
    n = len(df)
    cutoff = max(0, n - lag)   # 最后 lag 行数据窗口未关闭
    xs, vs, ds = [], [], []
    for i in range(cutoff):
        v = df[col].iloc[i]
        if pd.notna(v) and v > 0:
            xs.append(i); vs.append(float(v)); ds.append(df["日期标签"].iloc[i])
    return xs, vs, ds


def weighted_avg_7d(df, col, lag):
    """近7天完整数据加权均值，返回 (avg, d_start, d_end, n_valid)
    ★ v6：若数据行数不足，自动缩短窗口而非报错
    """
    n = len(df)
    lci = n - lag - 1          # 最后一个窗口已关闭的行索引
    if lci < 0:
        return np.nan, "N/A", "N/A", 0
    s0 = max(0, lci - 6); s1 = lci
    sub_r = df[col].iloc[s0:s1 + 1]
    sub_f = df["首充人数"].iloc[s0:s1 + 1]
    sub_d = df["日期标签"].iloc[s0:s1 + 1]
    mask = (sub_r > 0) & sub_r.notna() & sub_f.notna() & (sub_f > 0)
    r, f, d = sub_r[mask], sub_f[mask], sub_d[mask]
    if f.sum() == 0:
        return np.nan, "N/A", "N/A", 0
    return (float((r * f).sum() / f.sum()),
            d.iloc[0] if len(d) else "N/A",
            d.iloc[-1] if len(d) else "N/A",
            int(mask.sum()))


# ══════════════════════════════════════════════════════════════════════════════
# 3. 图表工具
# ══════════════════════════════════════════════════════════════════════════════

def _fig(w=16, h=5):
    fig, ax = plt.subplots(figsize=(w, h), facecolor="white")
    ax.set_facecolor("white")
    ax.spines[["top", "right"]].set_visible(False)
    ax.spines[["left", "bottom"]].set_color("#D8CEBD")
    ax.tick_params(colors=C_MUTED, labelsize=7)
    return fig, ax


def _label_bars(ax, bars, fmt="{:.0f}", size=7, color=C_DARK, offset=3):
    for bar in bars:
        h = bar.get_height()
        if h and not np.isnan(h) and abs(h) > 0:
            ax.text(bar.get_x() + bar.get_width() / 2, h + offset,
                    fmt.format(h), ha="center", va="bottom",
                    fontsize=size, color=color, fontproperties=_FP,
                    bbox=dict(boxstyle="round,pad=0.1", fc="white", ec="none", alpha=0.8))


def _label_line(ax, xs, ys, fmt="{:.1f}", size=7, color=C_DARK, offset=8, skip=4):
    for i, (x, y) in enumerate(zip(xs, ys)):
        if y is None or (isinstance(y, float) and np.isnan(y)): continue
        if i % skip != 0: continue
        ax.annotate(fmt.format(y), (x, y), textcoords="offset points", xytext=(0, offset),
                    ha="center", va="bottom", fontsize=size, color=color,
                    fontproperties=_FP,
                    bbox=dict(boxstyle="round,pad=0.15", fc="white", ec=color, alpha=0.88, lw=0.5))


def _to_img(fig, dpi=150) -> io.BytesIO:
    buf = io.BytesIO()
    fig.savefig(buf, format="png", dpi=dpi, bbox_inches="tight",
                facecolor="white", edgecolor="none")
    plt.close(fig); buf.seek(0); return buf


def _x(df):  return list(range(len(df)))
def _lb(df): return df["日期标签"].tolist()


def _xticks(ax, df):
    """★ v6：步长完全自适应实际数据量，保证不超过 20 个刻度标签"""
    xs = _x(df); lb = _lb(df)
    step = max(1, len(xs) // 20)
    ax.set_xticks(xs[::step])
    ax.set_xticklabels(lb[::step], fontproperties=_FP, fontsize=7)


# ══════════════════════════════════════════════════════════════════════════════
# 4. 图表定义
# ══════════════════════════════════════════════════════════════════════════════

def chart_dep_cost(df):
    """日充值 vs 日真实消耗（消耗排除最新一天）"""
    fig, ax = _fig(16, 4.5); ax2 = ax.twinx(); xs = _x(df)
    bars = ax.bar(xs, df["充值金额"].tolist(), color=C_GOLD, alpha=0.7, width=0.6)
    _label_bars(ax, bars, fmt="{:,.0f}", size=7, offset=5)
    ax2.plot(xs, df["真实消耗"].tolist(), color=C_RED, lw=1.8, marker="o", ms=3)
    _label_line(ax2, xs, df["真实消耗"].tolist(), fmt="{:,.0f}", size=7, color=C_RED, skip=4)
    _xticks(ax, df)
    ax.set_ylabel("充值金额", fontproperties=_FP, fontsize=8)
    ax2.set_ylabel("真实消耗", fontproperties=_FP, fontsize=8)
    ax2.spines[["top"]].set_visible(False)
    ax.set_title("日充值金额 vs 日真实消耗（消耗已排除最新一天未完整数据）",
                 fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend([Patch(color=C_GOLD, alpha=0.7), Line2D([0], [0], color=C_RED, lw=2, marker='o')],
              ["充值金额", "真实消耗"], prop=_FP, fontsize=8, loc="upper left", framealpha=0.7)
    fig.tight_layout(); return _to_img(fig)


def chart_diff(df):
    fig, ax = _fig(16, 4.5); ax2 = ax.twinx(); xs = _x(df)
    bars = []
    for x, v in zip(xs, df["充提差"].tolist()):
        if pd.notna(v):
            b = ax.bar(x, v, color=C_GREEN if v >= 0 else C_RED, alpha=0.7, width=0.6)
            bars.append(b[0])
    _label_bars(ax, bars, fmt="{:,.0f}", size=7)
    ax2.plot(xs, df["充提差比"].tolist(), color=C_GOLD, lw=1.8, marker="D", ms=3)
    _label_line(ax2, xs, df["充提差比"].tolist(), fmt="{:.1f}%", size=7, color=C_GOLD, skip=4)
    _xticks(ax, df)
    ax.set_ylabel("充提差", fontproperties=_FP, fontsize=8)
    ax2.set_ylabel("充提差比%", fontproperties=_FP, fontsize=8)
    ax2.spines["top"].set_visible(False)
    ax.set_title("充提差 & 充提差比", fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.axhline(0, color="#CCCCCC", lw=0.8, ls="--")
    fig.tight_layout(); return _to_img(fig)


def chart_roi(df):
    fig, ax = _fig(16, 5); xs = _x(df)
    dr  = df["充提差比"].tolist()
    roi = df["累计毛利ROI"].tolist()
    ax.fill_between(xs, dr, alpha=0.12, color=C_GOLD)
    ax.plot(xs, dr,  color=C_GOLD,  lw=1.8, marker="o", ms=3, label="当日充提差比%")
    _label_line(ax, xs, dr,  fmt="{:.1f}%", size=7, color=C_GOLD, skip=4)
    ax.plot(xs, roi, color=C_GREEN, lw=2, ls="--", marker="s", ms=3, label="累计毛利ROI%")
    _label_line(ax, xs, roi, fmt="{:.1f}%", size=7, color=C_GREEN, offset=10, skip=4)
    _xticks(ax, df)
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter("%.1f%%"))
    ax.set_title("充提差比 vs 累计毛利ROI（历史累计毛利÷历史累计真实消耗）",
                 fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend(prop=_FP, fontsize=8, framealpha=0.7, loc="upper left")
    fig.tight_layout(); return _to_img(fig)


def chart_cum_profit(df):
    fig, ax = _fig(16, 5); ax2 = ax.twinx(); xs = _x(df)
    bars = ax.bar(xs, df["历史累计真实消耗"].tolist(), color=C_MUTED, alpha=0.5, width=0.6)
    ax2.plot(xs, df["历史累计毛利"].tolist(), color=C_RED, lw=2, marker="o", ms=3)
    _label_line(ax2, xs, df["历史累计毛利"].tolist(), fmt="{:,.0f}", size=7, color=C_RED, skip=4)
    _xticks(ax, df)
    ax.set_ylabel("历史累计真实消耗", fontproperties=_FP, fontsize=8)
    ax2.set_ylabel("历史累计毛利", fontproperties=_FP, fontsize=8)
    ax2.spines["top"].set_visible(False)
    ax.set_title("历史累计真实消耗 & 历史累计毛利", fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend([Patch(color=C_MUTED, alpha=0.5), Line2D([0], [0], color=C_RED, lw=2)],
              ["历史累计真实消耗", "历史累计毛利"], prop=_FP, fontsize=8, loc="upper left", framealpha=0.7)
    fig.tight_layout(); return _to_img(fig)


def chart_arppu(df):
    fig, ax = _fig(16, 4.5); xs = _x(df)
    ax.fill_between(xs, df["人均充值金额"].tolist(), alpha=0.10, color=C_GREEN)
    ax.plot(xs, df["人均充值金额"].tolist(), color=C_GREEN, lw=1.8, marker="o", ms=3, label="全量ARPPU")
    _label_line(ax, xs, df["人均充值金额"].tolist(), fmt="${:.2f}", size=7, color=C_GREEN, skip=4)
    ax.plot(xs, df["老用户ARPPU"].tolist(), color=C_DARK, lw=1.8, ls="--", marker="s", ms=3, label="老用户ARPPU")
    _label_line(ax, xs, df["老用户ARPPU"].tolist(), fmt="${:.2f}", size=7, color=C_DARK, offset=10, skip=4)
    _xticks(ax, df)
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter("$%.1f"))
    ax.set_title("老用户ARPPU vs 全量ARPPU", fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend(prop=_FP, fontsize=8, framealpha=0.7, loc="upper left")
    fig.tight_layout(); return _to_img(fig)


def chart_payrate(df):
    fig, ax = _fig(16, 5); xs = _x(df)
    ax.plot(xs, df["新充付费率"].tolist(), color=C_BLUE, lw=1.8, marker="o",  ms=3, label="新充付费率%（新充/注册）")
    ax.plot(xs, df["老充付费率"].tolist(), color=C_GREEN, lw=1.8, ls="--", marker="s", ms=3, label="老充付费率%")
    ax.plot(xs, df["首充转化率"].tolist(), color=C_GOLD, lw=1.8, ls=":", marker="^", ms=3, label="首充转化率%")
    _xticks(ax, df)
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter("%.1f%%"))
    ax.set_title("新充付费率 vs 老充付费率 & 首充转化率", fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend(prop=_FP, fontsize=8, framealpha=0.7, loc="upper left")
    fig.tight_layout(); return _to_img(fig)


def chart_users(df):
    fig, ax = _fig(16, 4.5); xs = _x(df); w = 0.35
    b1 = ax.bar([x - w/2 for x in xs], df["注册人数"].tolist(), width=w, color=C_BLUE,  alpha=0.65, label="注册人数")
    b2 = ax.bar([x + w/2 for x in xs], df["首充人数"].tolist(), width=w, color=C_GOLD, alpha=0.85, label="首充人数")
    _label_bars(ax, b1, fmt="{:,.0f}", size=7, offset=3)
    _label_bars(ax, b2, fmt="{:,.0f}", size=7, offset=3)
    _xticks(ax, df)
    ax.set_title("首充人数 & 注册人数", fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend(prop=_FP, fontsize=8, framealpha=0.7, loc="upper left")
    fig.tight_layout(); return _to_img(fig)


def chart_cac_bonus(df):
    """一级首充获客成本 & 总赠送充值比（获客成本排除最新一天未完整数据）
    ★ v6：字段名改为 CAC_COL（一级首充获客成本），若列不存在则跳过
    """
    fig, ax = _fig(16, 4.5); ax2 = ax.twinx(); xs = _x(df)

    # ★ 兼容字段不存在的情况
    if CAC_COL in df.columns:
        cac = [float(r[CAC_COL]) if pd.notna(r.get(CAC_COL)) else None
               for _, r in df.iterrows()]
    else:
        log.warning(f"列「{CAC_COL}」不存在，获客成本图将留空")
        cac = [None] * len(df)

    bonus = df["总赠送充值比"].tolist()
    ax.plot(xs, cac, color=C_DARK, lw=1.8, marker="o", ms=4, label="获客成本$", zorder=3)
    _label_line(ax, xs, cac, fmt="${:.1f}", size=7, color=C_DARK, skip=4)
    bars = ax2.bar(xs, bonus, color=C_RED, alpha=0.4, width=0.55, label="赠送充值比%")
    _label_bars(ax2, bars, fmt="{:.1f}%", size=7, color=C_RED)
    _xticks(ax, df)
    ax.set_ylabel("获客成本$", fontproperties=_FP, fontsize=8)
    ax2.set_ylabel("赠送充值比%", fontproperties=_FP, fontsize=8)
    ax2.spines["top"].set_visible(False)
    ax.set_title(f"一级首充获客成本 & 总赠送充值比（获客成本已排除最新一天未完整数据）",
                 fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend([Line2D([0], [0], color=C_DARK, lw=2, marker='o'), Patch(color=C_RED, alpha=0.4)],
              ["获客成本$", "赠送充值比%"], prop=_FP, fontsize=8, loc="upper left", framealpha=0.7)
    fig.tight_layout(); return _to_img(fig)


def chart_bet_game(df):
    fig, ax = _fig(16, 4.5); ax2 = ax.twinx(); xs = _x(df)
    bet  = (df["投注金额"] / 1000).tolist()
    game = df["游戏盈利率"].tolist()
    bars = ax.bar(xs, bet, color=C_MUTED, alpha=0.55, width=0.6)
    _label_bars(ax, bars, fmt="{:.0f}K", size=7, offset=5)
    ax2.plot(xs, game, color=C_RED, lw=1.8, marker="D", ms=3)
    _label_line(ax2, xs, game, fmt="{:.1f}%", size=7, color=C_RED, skip=4)
    _xticks(ax, df)
    ax.set_ylabel("投注额(K)", fontproperties=_FP, fontsize=8)
    ax2.set_ylabel("游戏盈利率%", fontproperties=_FP, fontsize=8)
    ax2.spines["top"].set_visible(False)
    ax.set_title("日投注金额（千USD）& 游戏盈利率", fontproperties=_FP, fontsize=10, color=C_DARK)
    ax.legend([Patch(color=C_MUTED, alpha=0.55), Line2D([0], [0], color=C_RED, lw=2, marker='D')],
              ["投注额(K)", "游戏盈利率%"], prop=_FP, fontsize=8, loc="upper left", framealpha=0.7)
    fig.tight_layout(); return _to_img(fig)


# ─── ★ 留存趋势图（图例外置，不重叠）────────────────────────────────────────
def chart_repay_trend(df):
    """
    图例外置方案：
    - figsize=(17,6) 右侧多出1英寸空白
    - bbox_to_anchor=(1.01,1) 将图例框放在绘图区右侧外
    - subplots_adjust(right=0.82) 为图例预留空间
    - 不达标点用红色 × 符号
    ★ v6：get_complete_series 已自适应数据量，此处无需额外改动
    """
    SERIES = [
        ("首充2日复充率",  "2日复充",  21.0, C_BLUE,   "-",   "o",  1),
        ("首充3日复充率",  "3日复充",  15.0, C_GREEN,  "--",  "s",  2),
        ("首充7日复充率",  "7日复充",  11.0, C_GOLD,   ":",   "^",  6),
        ("首充14日复充率", "14日复充",  8.0, C_RED,    "-.",  "D", 13),
        ("首充30日复充率", "30日复充",  6.0, C_ORANGE, "--",  "P", 29),
    ]
    fig, ax = plt.subplots(figsize=(17, 6), facecolor="white")
    ax.set_facecolor("white")
    ax.spines[["top", "right"]].set_visible(False)
    ax.spines[["left", "bottom"]].set_color("#D8CEBD")
    ax.tick_params(colors=C_MUTED, labelsize=7)

    for col, lbl, target, clr, ls_, mk, lag in SERIES:
        if col not in df.columns: continue
        xs_ok, vals_ok, _ = get_complete_series(df, col, lag)
        if not xs_ok: continue
        ax.plot(xs_ok, vals_ok, color=clr, lw=1.8, ls=ls_, marker=mk, ms=3.5,
                label=f"{lbl}（目标{target:.0f}%）", zorder=3)
        miss_xs   = [x for x, v in zip(xs_ok, vals_ok) if v < target]
        miss_vals = [v for v in vals_ok if v < target]
        if miss_xs:
            ax.scatter(miss_xs, miss_vals, marker='x', color=C_RED,
                       s=55, zorder=6, linewidths=1.8)
        ax.plot([xs_ok[0], xs_ok[-1]], [target, target],
                color=clr, lw=0.9, ls=":", alpha=0.45, zorder=2)
        ax.annotate(f"目标{target:.0f}%",
                    xy=(xs_ok[-1], target), xytext=(4, 0),
                    textcoords="offset points",
                    ha="left", va="center", fontsize=7.5,
                    fontproperties=_FP, color=clr, alpha=0.9,
                    bbox=dict(boxstyle="round,pad=0.12", fc="white",
                              ec=clr, alpha=0.85, lw=0.6))

    xs_all = _x(df); lb_all = _lb(df)
    step = max(1, len(xs_all) // 20)
    ax.set_xticks(xs_all[::step])
    ax.set_xticklabels(lb_all[::step], fontproperties=_FP, fontsize=7)
    ax.yaxis.set_major_formatter(mticker.FormatStrFormatter("%.0f%%"))
    ax.set_ylim(bottom=0)
    ax.set_title("首充复充率趋势（2/3/7/14/30日）— 仅完整观测数据，红×=不达标",
                 fontproperties=_FP, fontsize=10, color=C_DARK)

    hs, ls2 = ax.get_legend_handles_labels()
    hs.append(Line2D([0], [0], marker='x', color=C_RED, linestyle='none',
                     markersize=8, markeredgewidth=1.8, label="不达标"))
    ls2.append("不达标")
    ax.legend(hs, ls2, prop=_FP, fontsize=8.5, framealpha=0.92,
              loc="upper left",
              bbox_to_anchor=(1.01, 1.0),
              borderaxespad=0,
              edgecolor="#D8CEBD")
    fig.subplots_adjust(right=0.82)
    return _to_img(fig)


# ══════════════════════════════════════════════════════════════════════════════
# 5. PDF 构建
# ══════════════════════════════════════════════════════════════════════════════
W, H   = A4
MARGIN = 14 * mm
IW     = W - 2 * MARGIN


def build_pdf(df: pd.DataFrame, cfg: dict, pdf_path: Path):
    doc = SimpleDocTemplate(str(pdf_path), pagesize=A4,
        leftMargin=MARGIN, rightMargin=MARGIN,
        topMargin=14 * mm, bottomMargin=12 * mm,
        title=f"{cfg['name']} 大盘汇总表")

    S = getSampleStyleSheet()
    def ps(name, **kw):
        kw.setdefault("fontName", "NotoSC")
        return ParagraphStyle(name, parent=S["Normal"], **kw)

    s_title = ps("t",  fontSize=16, textColor=colors.HexColor(C_DARK), fontName="NotoSC-B", leading=20, spaceAfter=2)
    s_sub   = ps("sb", fontSize=9,  textColor=colors.HexColor(C_MUTED), spaceAfter=4)
    s_sec   = ps("sc", fontSize=11, textColor=colors.HexColor(C_DARK),  fontName="NotoSC-B", spaceBefore=8, spaceAfter=4, leftIndent=8)
    s_note  = ps("n",  fontSize=7.5, textColor=colors.HexColor(C_MUTED), spaceBefore=2, spaceAfter=2, leading=11)
    s_lbl   = ps("l",  fontSize=7.5, textColor=colors.HexColor(C_MUTED), leading=10)
    s_val   = ps("v",  fontSize=13, textColor=colors.HexColor(C_DARK),  fontName="NotoSC-B", leading=15)

    TBL = TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0),  colors.HexColor("#F5F0E8")),
        ("BACKGROUND", (0, 1), (-1, -1), colors.white),
        ("BOX",        (0, 0), (-1, -1), 0.5, colors.HexColor("#E2D9CC")),
        ("INNERGRID",  (0, 0), (-1, -1), 0.3, colors.HexColor("#EDE7DB")),
        ("TOPPADDING",    (0, 0), (-1, -1), 5),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 5),
        ("LEFTPADDING",   (0, 0), (-1, -1), 7),
        ("RIGHTPADDING",  (0, 0), (-1, -1), 7),
        ("VALIGN",     (0, 0), (-1, -1), "MIDDLE"),
    ])

    def kpi3(items, cw=None):
        nc = len(items); cw = cw or [IW / nc] * nc
        td = [[Paragraph(a, s_lbl) for a, _, _ in items],
              [Paragraph(str(b), s_val) for _, b, _ in items],
              [Paragraph(str(c) if c else "", s_note) for _, _, c in items]]
        t = Table(td, colWidths=cw, hAlign="LEFT"); t.setStyle(TBL); return t

    def sec(txt):
        return Paragraph(f'<font color="{C_GOLD}">▌</font> {txt}', s_sec)

    # ── 数据 ──────────────────────────────────────────────────
    n    = len(df)
    last = df.iloc[-1]
    valid_dep = df[df["充值金额"].fillna(0) > 0]
    today = valid_dep.iloc[-1] if len(valid_dep) >= 1 else last
    yest  = valid_dep.iloc[-2] if len(valid_dep) >= 2 else today

    # ==================================================
    # 日充提差ROI特殊处理
    # 最新一天消耗未完整
    # 取 D-1 VS D-2
    # ==================================================
    roi_today = np.nan
    roi_yest  = np.nan
    
    if len(df) >= 3:
        roi_today = df["日充提差ROI"].iloc[-2]   # 昨天
        roi_yest  = df["日充提差ROI"].iloc[-3]   # 前天
    elif len(df) >= 2:
        roi_today = df["日充提差ROI"].iloc[-2]

    # 累计字段：取最新行（保持原逻辑）
    hist_consume  = last["历史累计真实消耗"]
    hist_profit   = last["历史累计毛利"]
    cum_roi_str   = f"{hist_profit / hist_consume * 100:.1f}%" if hist_consume > 0 else "N/A"
    cum_fdc_total = int(df["首充人数"].fillna(0).sum())

    # ★ v6：均值窗口使用实际有效天数（消耗/获客成本已在load_df置NaN，dropna自动跳过最新一天）
    active = df[df["充值金额"].fillna(0) > 0]
    n_active = len(active)   # 实际有效天数，用于 KPI 注释

    def avgc(col):
        if col not in active.columns: return np.nan
        return active[col].dropna().mean()

    avg_consume = avgc("真实消耗")
    avg_deposit = avgc("充值金额")
    avg_fdc     = avgc("首充人数")
    avg_reg     = avgc("注册人数")

    # ★ v6：获客成本字段名改为 CAC_COL
    avg_cac = active[CAC_COL].dropna().mean() if CAC_COL in active.columns else np.nan

    r2_avg, r2_d0, r2_d1, _ = weighted_avg_7d(df, "首充2日复充率", 1)
    r2_str = f"{r2_avg:.1f}%" if not np.isnan(r2_avg) else "N/A"

    date_range = f"{df['日期标签'].iloc[0]} — {df['日期标签'].iloc[-1]}"
    now_str    = datetime.now().strftime("%Y.%m.%d %H:%M")

    CLR_UP = "#1A6636"; CLR_DN = "#8B1A14"

    def delta_p(col, inv=False):
        tv = today.get(col, np.nan); yv = yest.get(col, np.nan)
        if pd.isna(tv) or pd.isna(yv) or yv == 0: return Paragraph("─", s_note)
        d = tv - yv; pct = d / abs(yv) * 100
        arr = "▲" if d > 0 else "▼"; good = (d > 0) if not inv else (d < 0)
        c = CLR_UP if good else CLR_DN
        return Paragraph(f'<font color="{c}"><b>{arr}{abs(pct):.1f}%</b></font>', s_note)

    def yv_s(col, unit="", dec=1):
        v = yest.get(col, np.nan)
        if pd.isna(v): return "昨日 N/A"
        return f"昨日 {v:,.0f}{unit}" if dec == 0 else f"昨日 {v:.{dec}f}{unit}"

    def td_i(label, col, unit="", dec=1, inv=False):
        tv = today.get(col, np.nan)
        val = "N/A" if pd.isna(tv) else (f"{tv:,.0f}{unit}" if dec == 0 else f"{tv:.{dec}f}{unit}")
        return (label, val, yv_s(col, unit, dec), delta_p(col, inv))
    def td_roi():
            """
            日充提差ROI
            显示：
            昨日(D-1) VS 前天(D-2)
            """
        
            if pd.isna(roi_today):
                return (
                    "日充提差ROI（昨日VS前天）",
                    "N/A",
                    "前天 N/A",
                    Paragraph("─", s_note)
                )
        
            value = f"{roi_today:.1f}%"
        
            if pd.isna(roi_yest) or roi_yest == 0:
        
                return (
                    "日充提差ROI（昨日VS前天）",
                    value,
                    "前天 N/A",
                    Paragraph("─", s_note)
                )
        
            diff = roi_today - roi_yest
            pct  = diff / abs(roi_yest) * 100
        
            arrow = "▲" if diff > 0 else "▼"
        
            color = CLR_UP if diff > 0 else CLR_DN
        
            delta = Paragraph(
                f'<font color="{color}"><b>{arrow}{abs(pct):.1f}%</b></font>',
                s_note
            )
        
            return (
                "日充提差ROI（昨日VS前天）",
                value,
                f"前天 {roi_yest:.1f}%",
                delta
            )

    def kpi4(items, cw=None):
        nc = len(items); cw = cw or [IW / nc] * nc
        td = [[Paragraph(a, s_lbl) for a, _, _, _ in items],
              [Paragraph(str(b), s_val) for _, b, _, _ in items],
              [Paragraph(c, s_note) for _, _, c, _ in items],
              [d for _, _, _, d in items]]
        t = Table(td, colWidths=cw, hAlign="LEFT"); t.setStyle(TBL); return t

    def ret_summary_tbl():
        hdr    = ps("rh",  fontName="NotoSC-B", fontSize=8.5, textColor=colors.HexColor(C_DARK))
        cell   = ps("rc",  fontName="NotoSC",   fontSize=9,   textColor=colors.HexColor(C_DARK))
        note_c = ps("rnc", fontName="NotoSC",   fontSize=7.5, textColor=colors.HexColor(C_MUTED), leading=11)
        gn     = ps("rg",  fontName="NotoSC-B", fontSize=9,   textColor=colors.HexColor(C_GREEN))
        rd     = ps("rr",  fontName="NotoSC-B", fontSize=9,   textColor=colors.HexColor(C_RED))
        rows = [[Paragraph(t, hdr) for t in ["指标", "目标", "实际留存率（近7天）", "差距(pp)", "备注（数据区间）"]]]
        for label, col, target, lag in RETENTION_TARGETS:
            if col not in df.columns:
                rows.append([Paragraph(s, cell) for s in [label, f"{target:.0f}%", "N/A", "─", "─"]])
                continue
            avg, d0, d1, nd = weighted_avg_7d(df, col, lag)
            # ★ v6：数据不足时给出提示而非空白
            if nd == 0:
                remark = f"数据不足（行数{n} < lag{lag}+1）"
            else:
                remark = f"{d0}~{d1}（{nd}天）"
            if np.isnan(avg):
                rows.append([Paragraph(label, cell), Paragraph(f"{target:.0f}%", cell),
                             Paragraph("N/A", cell), Paragraph("─", cell), Paragraph(remark, note_c)])
            else:
                gap = avg - target; ok = gap >= 0; gs = gn if ok else rd
                rows.append([Paragraph(label, cell), Paragraph(f"{target:.0f}%", cell),
                             Paragraph(f"{avg:.1f}%", gs), Paragraph(f"{gap:+.1f}", gs),
                             Paragraph(remark, note_c)])
        cw = [IW * p for p in [0.18, 0.11, 0.22, 0.15, 0.34]]
        tbl = Table(rows, colWidths=cw, hAlign="LEFT")
        tbl.setStyle(TableStyle([
            ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#F5F0E8")),
            ("BOX",        (0, 0), (-1, -1), 0.5, colors.HexColor("#E2D9CC")),
            ("INNERGRID",  (0, 0), (-1, -1), 0.3, colors.HexColor("#EDE7DB")),
            ("TOPPADDING",    (0, 0), (-1, -1), 7),
            ("BOTTOMPADDING", (0, 0), (-1, -1), 7),
            ("LEFTPADDING",   (0, 0), (-1, -1), 8),
            ("RIGHTPADDING",  (0, 0), (-1, -1), 8),
            ("VALIGN",     (0, 0), (-1, -1), "MIDDLE"),
        ]))
        return tbl

    # ── Story ──────────────────────────────────────────────────
    story = []
    story.append(Paragraph(f"{cfg['name']} 大盘平台汇总表", s_title))
    story.append(Paragraph(
        f"数据区间：{date_range}（近{n}日）&nbsp;&nbsp;|&nbsp;&nbsp;生成：{now_str}", s_sub))
    story.append(HRFlowable(width=IW, thickness=1.5, color=colors.HexColor(C_GOLD), spaceAfter=6))

    # 累计汇总
    story.append(sec("累计汇总"))
    story.append(kpi3([
        ("历史累计真实消耗",   f"{hist_consume:,.0f}",  "取最新日累计字段"),
        ("历史累计毛利",       f"{hist_profit:,.0f}",   "取最新日累计字段"),
        ("累计毛利ROI",        cum_roi_str,              "历史累计毛利÷历史累计真实消耗"),
        ("累计首充人数",       f"{cum_fdc_total:,}",    "本期合计"),
        # ★ v6：注释显示实际有效天数（排除最新一天未完整）
        ("日均真实消耗",
         f"{avg_consume:,.0f}" if pd.notna(avg_consume) else "N/A",
         f"近{n_active}日均值（排除最新一天未完整）"),
    ]))
    story.append(Spacer(1, 4))
    story.append(kpi3([
        ("日均充值金额",           f"{avg_deposit:,.0f}" if pd.notna(avg_deposit) else "N/A", f"近{n_active}日有效均值"),
        ("日均首充人数",           f"{avg_fdc:,.0f}"     if pd.notna(avg_fdc)     else "N/A", f"近{n_active}日有效均值"),
        ("日均注册人数",           f"{avg_reg:,.0f}"     if pd.notna(avg_reg)     else "N/A", f"近{n_active}日有效均值"),
        # ★ v6：标签改为「一级首充获客成本」
        ("一级首充获客成本（均）", f"${avg_cac:.2f}"     if pd.notna(avg_cac)     else "N/A", "排除最新一天未完整"),
        ("次留均值（近7天）",      r2_str,                                                     f"2日复充 {r2_d0}~{r2_d1}"),
    ]))
    story.append(Spacer(1, 6))

    # 今日 vs 昨日
    story.append(sec("今日 vs 昨日"))
    story.append(kpi4([
        td_i("今日充值金额", "充值金额", dec=0),
        td_i("充提差",       "充提差",   dec=0),
        td_i("充提差比",     "充提差比", unit="%"),
        td_i("首充人数",     "首充人数", dec=0),
        td_i("注册人数",     "注册人数", dec=0),
    ]))
    story.append(Spacer(1, 4))
    story.append(kpi4([
        td_i("全量ARPPU",   "人均充值金额", unit="$", dec=2),
        # td_i("首充转化率",  "首充转化率",  unit="%"),
        td_i("新充付费率",  "新充付费率",  unit="%"),
        td_i("老充付费率",  "老充付费率",  unit="%"),
        td_i("游戏盈利率",  "游戏盈利率",  unit="%"),
        td_roi(),
    ]))
    story.append(Spacer(1, 6))

    # ★ 复充率趋势图（图例外置）
    story.append(sec("首充复充率趋势（2/3/7/14/30日）"))
    story.append(Paragraph(
        "仅显示观测窗口已关闭的完整数据（各指标截止日期不同）。"
        "目标：2日21% / 3日15% / 7日11% / 14日8% / 30日6%。红色 × = 当日不达标。",
        s_note))
    story.append(Spacer(1, 3))
    story.append(Image(chart_repay_trend(df), width=IW, height=IW * 6 / 17))
    story.append(Spacer(1, 8))

    # ★ 留存目标汇总表（KeepTogether）
    story.append(KeepTogether([
        sec("首充留存/复充目标完成情况（近7天）"),
        Spacer(1, 3),
        Paragraph(
            "实际留存率 = 近7天完整数据加权均值（∑各日复充率×首充人数 ÷ ∑首充人数）。"
            "各指标观测窗口不同，具体日期见备注列。差距 = 实际 − 目标（pp=百分点）。",
            s_note),
        Spacer(1, 3),
        ret_summary_tbl(),
    ]))
    story.append(Spacer(1, 10))

    def add_chart(buf, title, h=4.5):
        story.append(sec(title))
        story.append(Image(buf, width=IW, height=IW * h / 16))
        story.append(Spacer(1, 6))

    log.info(f"[{cfg['name']}] 生成常规图表...")
    add_chart(chart_dep_cost(df),   "日充值金额 vs 日真实消耗")
    add_chart(chart_diff(df),       "充提差 & 充提差比")
    add_chart(chart_roi(df),        "充提差比 vs 累计毛利ROI", h=5)
    add_chart(chart_cum_profit(df), "历史累计真实消耗 & 历史累计毛利", h=5)
    add_chart(chart_arppu(df),      "老用户ARPPU vs 全量ARPPU")
    add_chart(chart_payrate(df),    "新充付费率 vs 老充付费率 & 首充转化率", h=5)
    add_chart(chart_users(df),      "首充人数 & 注册人数")
    add_chart(chart_cac_bonus(df),  "一级首充获客成本 & 总赠送充值比")   # ★ v6 标题
    add_chart(chart_bet_game(df),   "日投注金额（千USD）& 游戏盈利率")

    story.append(HRFlowable(width=IW, thickness=0.5,
                             color=colors.HexColor("#E2D9CC"), spaceBefore=4))
    fn = ps("fn", fontSize=7.5, textColor=colors.HexColor(C_MUTED), leading=11)
    note_text = (
        "① 真实消耗/一级首充获客成本排除最新一天（当日录入未完整）；<br/>"   # ★ v6
        "② N日复充剔除数据不完整的日期数据。<br/>"
        "③ 留存趋势图仅显示窗口已关闭的完整数据，红×=不达标，图例置于右侧外。<br/>"
        "④ 留存汇总表取近7天完整数据加权均值，随每日数据自动更新，日期见备注列。<br/>"
        "⑤ 日充提差ROI=充提差/真实消耗，因为当天真实消耗未录完，使用昨日VS前天。<br/>"
        "⑥ 时间轴从旧到新，自动兼容升降序 Excel；数据不足60天时自适应处理。"  # ★ v6
    )
    story.append(Paragraph(note_text, fn))

    log.info(f"[{cfg['name']}] 构建 PDF ...")
    doc.build(story)
    log.info(f"[{cfg['name']}] ✅ PDF 保存：{pdf_path}")


# ══════════════════════════════════════════════════════════════════════════════
# 6. Telegram 推送
# ══════════════════════════════════════════════════════════════════════════════

def send_pdf(pdf_path: Path, cfg: dict) -> bool:
    caption = (f"📊 *{cfg['name']} 大盘汇总表*\n"
               f"📅 {datetime.now().strftime('%Y-%m-%d %H:%M')}")
    try:
        with open(pdf_path, "rb") as f:
            r = requests.post(
                f"https://api.telegram.org/bot{cfg['bot_token']}/sendDocument",
                data={"chat_id": cfg["chat_id"], "caption": caption,
                      "parse_mode": "Markdown"},
                files={"document": (pdf_path.name, f, "application/pdf")},
                timeout=120)
        ok = r.json().get("ok", False)
        log.info(f"[{cfg['name']}] TG " + ("✅ 成功" if ok else f"❌ {r.json().get('description')}"))
        return ok
    except Exception as e:
        log.error(f"[{cfg['name']}] ❌ 发送异常：{e}")
        return False


# ══════════════════════════════════════════════════════════════════════════════
# 7. 主流程
# ══════════════════════════════════════════════════════════════════════════════

def run_platform(cfg: dict):
    log.info(f"\n{'=' * 55}")
    log.info(f"  平台：{cfg['name']}")
    log.info(f"{'=' * 55}")
    try:
        path = find_excel(cfg)
        df   = load_df(path)
        log.info(f"[{cfg['name']}] {df['日期标签'].iloc[0]} → {df['日期标签'].iloc[-1]}（{len(df)}日）")
        Path(cfg["output_dir"]).mkdir(parents=True, exist_ok=True)
        latest_date = str(df["日期"].iloc[-1])[:8]
        pdf_path = Path(cfg["output_dir"]) / f"{cfg['name']}_大盘汇总表_{latest_date}.pdf"
        build_pdf(df, cfg, pdf_path)
        send_pdf(pdf_path, cfg)
    except Exception as e:
        log.error(f"[{cfg['name']}] ❌ 失败：{e}", exc_info=True)


def main():
    log.info(f"\n{'=' * 55}")
    log.info(f"  多平台大盘日报 v6  {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    log.info(f"{'=' * 55}")
    for cfg in PLATFORMS:
        run_platform(cfg)
    log.info("\n✅ 全部完成")


if __name__ == "__main__":
    main()
```
